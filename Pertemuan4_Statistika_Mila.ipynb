```python
#tabel data
import pandas as pd
path = '/content/MilaAzizah.xlsx'
dataraw = pd.read_excel(path)
print(dataraw)
```

```python
#Tabel Frekuensi
datafrq = pd.crosstab(index=dataraw["Grade"], columns="Frekuensi")
print(datafrq)
```

```python
#Grafik Garis
datafrq.plot(color='green')
```

```python
#Grafik Batang
datafrq.plot(kind='bar', color='green')
```

```python
import matplotlib.pyplot as plt
warna_pastel = ["#ff9999", "#66b3ff", "#99ff99", "#ffcc99"]

datafrq.plot(
    kind="pie",
    y="Frekuensi",
    autopct="%.1f%%",
    legend=False,
    colors=warna_pastel,
)

plt.show()
```

```python
#Statistika Deskriptif
dataraw["Final Score"] = pd.to_numeric(dataraw["Final Score"],errors='coerce')
dt = dataraw["Final Score"]
stats = dt.describe()
stats['Standard Error'] = dt.sem()
stats['Median'] = dt.median()
stats['Mode'] = dt.mode().iloc[0]
stats['variance'] = dt.var()
stats['range'] = dt.max() - dt.min()
stats['skewness'] = dt.skew()
stats['kurtosis'] = dt.kurtosis()
print(stats)
```
